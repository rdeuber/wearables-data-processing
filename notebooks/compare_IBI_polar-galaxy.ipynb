```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
ppg_galaxy_detection = pd.read_csv('data/galaxy_processed_charlton.csv')
ppg_galaxy_detection = ppg_galaxy_detection[ppg_galaxy_detection['PPG_Peaks'] == 1]
ppg_galaxy_detection["datetime"] = pd.to_datetime(ppg_galaxy_detection["datetime"])
ppg_galaxy_detection['ibi'] = ppg_galaxy_detection["datetime"].diff(periods=-1).dt.total_seconds() * -1000
ppg_galaxy_detection['hr'] = 60000 / ppg_galaxy_detection["ibi"]

ppg_galaxy_detection
```

```python
ppg_polar_detection = pd.read_csv('data/polar_processed_charlton.csv')
ppg_polar_detection = ppg_polar_detection[ppg_polar_detection['PPG_Peaks'] == 1]
ppg_polar_detection["datetime"] = pd.to_datetime(ppg_polar_detection["datetime"])
ppg_polar_detection['ibi'] = ppg_polar_detection["datetime"].diff(periods=-1).dt.total_seconds() * -1000
ppg_polar_detection['hr'] = 60000 / ppg_polar_detection["ibi"] 


ppg_polar_detection
```

```python
plt.figure(figsize=[15, 8])
plt.plot(ppg_polar_detection["datetime"], ppg_polar_detection["ibi"], label = "polar", c="red", alpha=0.5)
plt.plot(ppg_galaxy_detection["datetime"], ppg_galaxy_detection["ibi"], label = "galaxy", c="blue", alpha=0.5)
plt.grid()
plt.legend()
```

```python
plt.figure(figsize=[15, 8])
plt.ylim([20, 180])
plt.plot(ppg_polar_detection["datetime"], ppg_polar_detection["hr"], label = "polar", c="red", alpha=0.5)
plt.plot(ppg_galaxy_detection["datetime"], ppg_galaxy_detection["hr"], label = "galaxy", c="blue", alpha=0.5)
plt.legend()
```
